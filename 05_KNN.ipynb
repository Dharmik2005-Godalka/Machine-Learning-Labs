```python
from sklearn.datasets import load_iris
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data=load_iris()
#entire dataset will be loaded in the the variable data
```

```python
data
```

```python
X=data.data
Y=data.target
```

```python
def euclidean_distance(X):
    dict1={}
    for i in range(1,150):
        distance=((X[0][0]-X[i][0])**2+(X[0][1]-X[i][1])**2+(X[0][2]-X[i][2])**2+(X[0][3]-X[i][3])**2) ** 0.5
        dict1[i]=distance
    return dict1
```

```python
result=euclidean_distance(X)
print(result)
```

```python
sorted_by_values=dict(sorted(result.items(),key=lambda item:item[1]))
```

```python
sorted_by_values
```

```python
k=10
i=0
for i in range(1,k+1):
    key_list=list(sorted_by_values)[:k]
print(key_list)
```

```python
target_list=[]
for i in key_list: #this gives us values of key_list
    target_list.append(Y[i+1])
print(target_list)
```

```python
unique, counts = np.unique(target_list, return_counts=True)
max_count = counts.max()
most_repeated = unique[counts == max_count]
print(most_repeated)
if len(most_repeated)!=1:
    print("Change the value of k as the element belongs to not just one class right now")
```

```python
if most_repeated==0:
    print("Belongs to class Setosa")
elif most_repeated==1:
    print("Belongs to class versicolor")
else:
    print("Belongs to class virginca")
```
